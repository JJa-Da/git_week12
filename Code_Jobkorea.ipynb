```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd

options = webdriver.ChromeOptions()
options.add_argument("--headless=new")

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

url = "https://www.jobkorea.co.kr/Search/?stext=데이터분석"
driver.get(url)

wait = WebDriverWait(driver, 10)

# 🔥 공고 박스가 로드되기를 기다림
items = wait.until(
    EC.presence_of_all_elements_located(
        (By.CSS_SELECTOR, "div[data-sentry-component='CardJob']")
    )
)

rows = []

for item in items:

    # 회사명
    try:
        company = item.find_element(By.CSS_SELECTOR,
            "a[rel='noopener noreferrer'] span[data-accent-color]"
        ).text.strip()
    except:
        company = ""

    # 제목
    try:
        title = item.find_element(By.CSS_SELECTOR,
            "a[data-sentry-component='BaseLink'] span[data-accent-color]"
        ).text.strip()
    except:
        title = ""

    # URL
    try:
        detail_url = item.find_element(By.CSS_SELECTOR,
            "a[href*='GI_Read']"
        ).get_attribute("href")
    except:
        detail_url = ""

    # 지역
    try:
        location = item.find_element(By.CSS_SELECTOR,
            "div[data-sentry-component='GrayChip'] span"
        ).text.strip()
    except:
        location = ""

    # 경력
    try:
        experience = item.find_element(By.CSS_SELECTOR,
            "span[data-accent-color='gray700']"
        ).text.strip()
    except:
        experience = ""

    rows.append({
        "site": "JobKorea",
        "Col_Company": company,
        "Col_Recruit": title,
        "Col_detail": f"{location}, {experience}",
        "Col_url": detail_url
    })

driver.quit()
df.to_csv("data_tmp/data_jobkorea.csv", index=False, encoding="utf-8-sig")
df = pd.DataFrame(rows)
df
```
