```python
from selenium import webdriver
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
import os
import time

options = webdriver.ChromeOptions()
options.add_argument("--headless=new")

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)

url = "https://www.jobkorea.co.kr/Search/?stext=데이터분석"
driver.get(url)

wait = WebDriverWait(driver, 10)


items = wait.until(
    EC.presence_of_all_elements_located(
        (By.CSS_SELECTOR, "div[data-sentry-component='CardJob']")
    )
)

rows = []

for item in items:
    
    try:
        recruit = item.find_element(
            By.CSS_SELECTOR,
            "a[data-sentry-element='BaseLink'] span.Typography_variant_size18__344nw25"
        ).text.strip()
    except:
        recruit = ""

    
    try:
        company = item.find_element(
            By.CSS_SELECTOR,
            "a[data-sentry-element='BaseLink'] span[data-accent-color='gray700']"
        ).text.strip()
    except:
        company = ""

   
    try:
        detail_url = item.find_element(
            By.CSS_SELECTOR,
            "a[data-sentry-element='BaseLink']"
        ).get_attribute("href")
    except:
        detail_url = ""

    
    try:
        chips = item.find_elements(
            By.CSS_SELECTOR,
            "div[data-sentry-component='GrayChip'] span"
        )
        detail = " | ".join([c.text.strip() for c in chips])
    except:
        detail = ""

    rows.append({
        "site": "jobkorea",
        "Col_Company": company,
        "Col_Recruit": recruit,
        "Col_detail": detail,
        "Col_url": detail_url
    })

driver.quit()

df = pd.DataFrame(rows)

os.makedirs("data_tmp", exist_ok=True)
df.to_csv("data_tmp/data_jobkorea.csv", index=False, encoding="utf-8-sig")

df
```
